```python
#for reproducibility
SEED = 60
np.random.seed(SEED)
tf.random.set_seed(SEED)
random.seed(SEED)
```

```python
#Dropping the 190 records without performance ratings from merged_df in preparation for Machine Learning
merged_df_cleaned = merged_df_latest.dropna().copy()
merged_df_cleaned.columns
```

```python
print(f"\n  Dataset shape   : {merged_df_cleaned.shape}")
print(f"  Missing values  : {merged_df_cleaned.isnull().sum().sum()}")
print(f"\n  Target Distribution:")
print(merged_df_latest["Attrition"].value_counts())
print(f"\n  Attrition Rate  : {(merged_df_cleaned['Attrition']=='Yes').mean()*100:.1f}% ")
```

```python
#Removing redundant columns to reduce dimensionality
columns_to_drop = [
    'FirstName', 'LastName', 'EmployeeID', 'PerformanceID',
    'ReviewDate', 'HireDate', 'EducationLevelID', 'SelfRating',
    'ManagerRating', 'Attrition_Rate','Attr_Rate','State'
]

columns_to_drop_existing = [col for col in columns_to_drop if col in merged_df_cleaned.columns]
merged_df_processed = merged_df_cleaned.drop(columns=columns_to_drop_existing)
merged_df_processed.shape
```

```python
satisfaction_mapping = {
    'Very Dissatisfied': 0,
    'Dissatisfied': 1,
    'Neutral': 2,
    'Satisfied':3,
    'Very Satisfied':4
}

satisfaction_cols = ['EnvironmentSatisfaction', 'JobSatisfaction', 'RelationshipSatisfaction', 'WorkLifeBalance']

for col in satisfaction_cols:
    if col in merged_df_processed.columns:
        merged_df_processed[col] = merged_df_processed[col].map(satisfaction_mapping)

print(merged_df_processed[satisfaction_cols].head())
```

```python
# Identify categorical columns for one-hot encoding (excluding the target 'Attrition')
categorical_cols = merged_df_processed.select_dtypes(include='object').columns.tolist()
if 'Attrition' in categorical_cols:
    categorical_cols.remove('Attrition')

# Exclude already ordinally encoded satisfaction columns from one-hot encoding
satisfaction_cols = ['EnvironmentSatisfaction', 'JobSatisfaction', 'RelationshipSatisfaction', 'WorkLifeBalance'] # Ensure this list is consistent with the ordinal mapping
categorical_cols = [col for col in categorical_cols if col not in satisfaction_cols]

# Apply one-hot encoding
merged_df_encoded = pd.get_dummies(merged_df_processed, columns=categorical_cols, drop_first=True)
```

```python
# Separate features (X) and target (y)
X = merged_df_encoded.drop('Attrition', axis=1)
y = merged_df_encoded['Attrition']

print(f"X shape: {X.shape}")
print(f"y shape: {y.shape}")
```

```python
#split the train and testing data of the merged_df_encoded
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=SEED)
print(f"\n  Train size: {X_train.shape[0]} | Test size: {X_test.shape[0]}")
```

```python
#Feature Scaling (fit on train, transform both)
# Initialize StandardScaler
scaler = StandardScaler()
X_train_sc = scaler.fit_transform(X_train)
X_test_sc  = scaler.transform(X_test)
```

```python
# Check the class distribution before SMOTE
print('Class distribution before SMOTE:')
print(y_train.value_counts())

#Apply SMOTE to training data to handle class imbalance
smote = SMOTE(random_state=SEED)
X_train_smote, y_train_smote = smote.fit_resample(X_train_sc, y_train)

# Check the class distribution after SMOTE
print('\nClass distribution after SMOTE:')
print(y_train_smote.value_counts())
```

```python
#**Logistic Regression**#
LogisticRegression_model = LogisticRegression(
        C=0.3, max_iter=3000, class_weight="balanced",
        solver="lbfgs", random_state=SEED)

#Training the Logistic Regression model
LogisticRegression_model.fit(X_train_smote, y_train_smote)
```

```python
# predict()       → class label  (0 = No Attrition, 1 = Attrition)
# predict_proba() → probability  ([P(Sta), P(Leave)])

y_pred = LogisticRegression_model.predict(X_test_sc)
y_prob = LogisticRegression_model.predict_proba(X_test_sc)[:, 1]   # probability of leaving

print(f"\nSample predictions (first 10):")
print(f"  Predicted labels      : {y_pred[:10].tolist()}")
print(f"  Predicted probability : {[round(p, 3) for p in y_prob[:10]]}")
print(f"  Actual labels         : {y_test.values[:10].tolist()}")
```

```python
from sklearn.metrics import f1_score

# Convert y_test to numerical labels to match y_pred_t
y_test_numeric = y_test.map({'No': 0, 'Yes': 1})

thresholds = np.arange(0.1, 10, 0.05)
f1_scores  = []

for t in thresholds:
    y_pred_t = (y_prob >= t).astype(int)
    f1_scores.append(f1_score(y_test_numeric, y_pred_t, zero_division=0))

best_threshold_Lr = thresholds[np.argmax(f1_scores)]
best_f1_Lr        = max(f1_scores)

print(f"Best threshold : {best_threshold_Lr:.2f}")
print(f"Best F1        : {best_f1_Lr:.4f}")
```

```python
custom_threshold = 0.8 # Define your custom threshold

# Re-calculate y_pred using the custom threshold
y_pred_custom = (y_prob >= custom_threshold).astype(int)

# Convert y_test to numerical labels to match y_pred_custom
y_test_numeric = y_test.map({'No': 0, 'Yes': 1})

#Logisitic regression model Evaluation with custom threshold
Accuracy  = accuracy_score(y_test_numeric, y_pred_custom)
Precision = precision_score(y_test_numeric, y_pred_custom, pos_label=1, zero_division=0)
Recall    = recall_score(y_test_numeric, y_pred_custom, pos_label=1, zero_division=0)
f1        = f1_score(y_test_numeric, y_pred_custom, pos_label=1, zero_division=0)

roc_auc   = roc_auc_score(y_test_numeric, y_prob) # ROC-AUC is threshold-independent

print("\n" + "=" * 45)
print("  LOGISTIC REGRESSION — EVALUATION RESULTS (Custom Threshold = {:.1f})".format(custom_threshold))
print("=" * 45)
print(f"  Accuracy  : {Accuracy:.4f}  ← misleading for imbalanced!")
print(f"  Precision : {Precision:.4f}")
print(f"  Recall    : {Recall:.4f}   ← how many leavers caught?")
print(f"  F1-Score  : {f1:.4f}   ← PRIMARY metric")
print(f"  ROC-AUC   : {roc_auc:.4f}   ← discrimination ability")
print("=" * 45)

print("\nDetailed Classification Report:")
print(classification_report(y_test_numeric, y_pred_custom,
      target_names=y.unique()))
```

### Binary Classification Neural Network (MLP)

```python
# Label Encode the Attrition data to 0 and 1 for Neural Network
label_encoder = LabelEncoder()
y_train_smote_encoded = label_encoder.fit_transform(y_train_smote)
y_test_encoded = label_encoder.transform(y_test)

print(f"Classes         : {label_encoder.classes_}")
print(f"Train encoded   : {np.unique(y_train_smote_encoded)}")
print(f"Test  encoded   : {np.unique(y_test_encoded)}")

print("Encoded y_train_smote unique values:", np.unique(y_train_smote_encoded))
print("Encoded y_test unique values:", np.unique(y_test_encoded))
```

```python
from sklearn.utils.class_weight import compute_class_weight

#Adding extra layer of robustiness by balancing class weights
classes = np.unique(y_train_smote_encoded)
weights = compute_class_weight("balanced",
                                classes=classes,
                                y=y_train_smote_encoded)
class_weight_dict = dict(zip(classes, weights))
```

```python
class F1ScoreCallback(keras.callbacks.Callback):
    def __init__(self, val_data, threshold=0.5):
        super().__init__()
        self.X_val      = val_data[0]
        self.y_val      = val_data[1]
        self.threshold  = threshold
        self.val_f1s    = []
        self.best_f1    = 0
        self.best_epoch = 0

    def on_epoch_end(self, epoch, logs=None):
        y_prob = self.model.predict(self.X_val, verbose=0).flatten()
        y_pred = (y_prob >= self.threshold).astype(int)
        f1     = f1_score(self.y_val, y_pred, zero_division=0)
        self.val_f1s.append(f1)
        logs['val_f1'] = f1
        if f1 > self.best_f1:
            self.best_f1    = f1
            self.best_epoch = epoch + 1
        if (epoch + 1) % 10 == 0:
            print(f"  Epoch {epoch+1:3d} — val_f1 = {f1:.4f}")
```

```python
# Define the neural network model for tabular binary classification
nn_model = keras.Sequential([
    # Input layer: Directly takes the features, specifying the input shape
    keras.layers.Dense(256, activation='relu', input_shape=(X_train_smote.shape[1],),
                       kernel_regularizer=keras.regularizers.l2(1e-4)),
    keras.layers.BatchNormalization(), # Batch normalization for stability
    keras.layers.Dropout(0.4, seed = SEED ), # Dropout for regularization
    keras.layers.Dense(128, activation='relu', input_shape=(X_train_smote.shape[1],),
                       kernel_regularizer=keras.regularizers.l2(1e-4)),
    keras.layers.BatchNormalization(), # Batch normalization for stability
    keras.layers.Dropout(0.35, seed = SEED ), # Dropout for regularization
    keras.layers.Dense(64, activation='relu', kernel_regularizer=keras.regularizers.l2(1e-4)), # Another hidden layer
    keras.layers.BatchNormalization(),
    keras.layers.Dropout(0.32, seed=SEED),
    keras.layers.Dense(32, activation='relu', kernel_regularizer=keras.regularizers.l2(1e-4)), # Another hidden layer
    keras.layers.Dropout(0.2, seed = SEED),
    keras.layers.Dense(1, activation='sigmoid') # Output layer for binary classification
])
#Compile the model
nn_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=8e-4),
    loss='binary_crossentropy', # Use binary_crossentropy for binary classification
    metrics=['accuracy', keras.metrics.AUC(name='auc'),  keras.metrics.Precision(name='precision'),
    keras.metrics.Recall(name='recall')]
)

nn_model.summary()
```

```python
# Use 15% of training data as validation (same split as validation_split=0.15)
val_size  = int(0.15 * len(X_train_smote))
X_val_cb  = X_train_smote[:val_size]
y_val_cb  = y_train_smote_encoded[:val_size]
X_tr_cb   = X_train_smote[val_size:]
y_tr_cb   = y_train_smote_encoded[val_size:]

f1_callback = F1ScoreCallback(val_data=(X_val_cb, y_val_cb), threshold=0.5)
```

```python
#Train the neural network
history = nn_model.fit(
    X_train_smote, y_train_smote_encoded, # Use scaled SMOTE data and encoded target
    epochs=150, # Increased epochs for better training
    batch_size=32,
    validation_split=0.15, # Use 10% of training data for validation
    class_weight=class_weight_dict,
    verbose=1,
    callbacks=[callbacks.EarlyStopping(monitor = 'val_loss', patience=20,
                                       restore_best_weights=True, verbose=1),
               callbacks.ReduceLROnPlateau(monitor = 'val_loss', factor = 0.5,
                                           patience = 8, min_lr=1e-6, verbose =1),
               callbacks.ModelCheckpoint(filepath='best_attrition_nn.keras',
                                         monitor='val_auc', save_best_only=True, mode='max',
                                         verbose=0)
               ],
    shuffle=True
)
```

```python
# Label Encode the target variable for Neural Network
label_encoder = LabelEncoder()
y_alt_train_smote_encoded = label_encoder.fit_transform(y_alt_train_smote)
y_alt_test_encoded = label_encoder.transform(y_alt_test)

print(f"Classes         : {label_encoder.classes_}")
print(f"Train encoded   : {np.unique(y_alt_train_smote_encoded)}")
print(f"Test  encoded   : {np.unique(y_alt_test_encoded)}")

print("Encoded y_train_smote unique values:", np.unique(y_alt_train_smote_encoded))
print("Encoded y_test unique values:", np.unique(y_alt_test_encoded))
```

```python
from sklearn.utils.class_weight import compute_class_weight

#Balancing the class weights for robustness
classes = np.unique(y_alt_train_smote_encoded)
weights = compute_class_weight("balanced",
                                classes=classes,
                                y=y_alt_train_smote_encoded)
class_weight_dict = dict(zip(classes, weights))
```

```python
#Defining custom keras callback
class F1ScoreCallback(keras.callbacks.Callback):
    def __init__(self, val_data, threshold=0.5):
        super().__init__()
        self.X_val      = val_data[0]
        self.y_val      = val_data[1]
        self.threshold  = threshold
        self.val_f1s    = []
        self.best_f1    = 0
        self.best_epoch = 0

    def on_epoch_end(self, epoch, logs=None):
        y_prob = self.model.predict(self.X_val, verbose=0).flatten()
        y_pred = (y_prob >= self.threshold).astype(int)
        f1     = f1_score(self.y_val, y_pred, zero_division=0)
        self.val_f1s.append(f1)
        logs['val_f1'] = f1
        if f1 > self.best_f1:
            self.best_f1    = f1
            self.best_epoch = epoch + 1
        if (epoch + 1) % 10 == 0:
            print(f"  Epoch {epoch+1:3d} — val_f1 = {f1:.4f}")
```

```python
# Define a new neural network model for the alternate dataset
# Ensure the input_shape matches the number of features in X_alt_train_smote
nn_model_alt = keras.Sequential([
    # Input layer: Directly takes the features, specifying the input shape
    keras.layers.Dense(256, activation='relu', input_shape=(X_alt_train_smote.shape[1],),
                       kernel_regularizer=keras.regularizers.l2(1e-4)),
    keras.layers.BatchNormalization(), # Batch normalization for stability
    keras.layers.Dropout(0.4, seed = SEED ), # Dropout for regularization
    keras.layers.Dense(128, activation='relu',
                       kernel_regularizer=keras.regularizers.l2(1e-4)),
    keras.layers.BatchNormalization(), # Batch normalization for stability
    keras.layers.Dropout(0.35, seed = SEED ), # Dropout for regularization
    keras.layers.Dense(64, activation='relu', kernel_regularizer=keras.regularizers.l2(1e-4)), # Another hidden layer
    keras.layers.BatchNormalization(),
    keras.layers.Dropout(0.32, seed=SEED),
    keras.layers.Dense(32, activation='relu', kernel_regularizer=keras.regularizers.l2(1e-4)), # Another hidden layer
    keras.layers.Dropout(0.2, seed = SEED),
    keras.layers.Dense(1, activation='sigmoid') # Output layer for binary classification
])

# Compile the new model
nn_model_alt.compile(
    optimizer=keras.optimizers.Adam(learning_rate=8e-4),
    loss='binary_crossentropy', # Use binary_crossentropy for binary classification
    metrics=['accuracy', keras.metrics.AUC(name='auc'),  keras.metrics.Precision(name='precision'),
    keras.metrics.Recall(name='recall')]
)

nn_model_alt.summary()
```

```python
# Use 15% of alternate training data as validation
alt_val_size  = int(0.15 * len(X_alt_train_smote))
X_alt_val_cb  = X_alt_train_smote[:alt_val_size]
y_alt_val_cb  = y_alt_train_smote_encoded[:alt_val_size]
X_alt_tr_cb   = X_alt_train_smote[alt_val_size:]
y_alt_tr_cb   = y_alt_train_smote_encoded[alt_val_size:]

alt_f1_callback = F1ScoreCallback(val_data=(X_alt_val_cb, y_alt_val_cb), threshold=0.5)
```

```python
history_alt = nn_model_alt.fit(
    X_alt_train_smote, y_alt_train_smote_encoded, # Use scaled SMOTE data and encoded target
    epochs=150,
    batch_size=32,
    validation_split=0.15,
    class_weight=class_weight_dict,
    verbose=1,
    callbacks=[callbacks.EarlyStopping(monitor = 'val_loss', patience=20,
                                       restore_best_weights=True, verbose=1),
               callbacks.ReduceLROnPlateau(monitor = 'val_loss', factor = 0.5,
                                           patience = 8, min_lr=1e-6, verbose =1),
               callbacks.ModelCheckpoint(filepath='best_attrition_nn_alt.keras',
                                         monitor='val_auc', save_best_only=True, mode='max',
                                         verbose=0),
               alt_f1_callback
               ],
    shuffle=True
)
```

```python
epochs_run_alt = len(history_alt.history['loss'])
best_epoch_alt = int(np.argmin(history_alt.history['val_loss'])) + 1
best_val_auc_alt = max(history_alt.history['val_auc'])

print(f"\n  Training stopped at epoch : {epochs_run_alt}")
print(f"  Best epoch (val_loss)     : {best_epoch_alt}")
print(f"  Best val_loss             : {min(history_alt.history['val_loss']):.4f}")
print(f"  Best val_auc              : {best_val_auc_alt:.4f}")
print(f"  Best val_f1 (custom CB)   : {alt_f1_callback.best_f1:.4f} at epoch {alt_f1_callback.best_epoch}")
```

```python
# Evaluate Neural Network for alternate data
loss_nn_alt, accuracy_nn_alt, auc_nn_alt, precision_nn_alt, recall_nn_alt = nn_model_alt.evaluate(X_alt_test_sc, y_alt_test_encoded, verbose=0)

# Make predictions
y_NNprob_alt = nn_model_alt.predict(X_alt_test_sc).flatten()
thresholds_nn_alt = np.arange(0.05, 0.65, 0.05)
f1_scores_nn_alt  = [f1_score(y_alt_test_encoded,
               (y_NNprob_alt >= t).astype(int),
               zero_division=0) for t in thresholds_nn_alt]
best_threshold_NN_alt = thresholds_nn_alt[np.argmax(f1_scores_nn_alt)]
best_f1_NN_alt        = max(f1_scores_nn_alt)
y_NNpred_alt   = (y_NNprob_alt >= best_threshold_NN_alt).astype(int)

print(f"\nSample predictions (first 10):")
print(f"  Predicted labels      : {y_NNpred_alt[:10].tolist()}")
print(f"  Predicted probability : {[round(p, 3) for p in y_NNprob_alt[:10]]}")
print(f"  Actual labels         : {y_alt_test_encoded[:10].tolist()}")

# Neural Network model Evaluation
Precision_nn_alt = precision_score(y_alt_test_encoded, y_NNpred_alt, zero_division=0)
Recall_nn_alt    = recall_score(y_alt_test_encoded, y_NNpred_alt, zero_division=0)
F1_nn_alt        = f1_score(y_alt_test_encoded, y_NNpred_alt, zero_division=0)
roc_auc_nn_alt   = roc_auc_score(y_alt_test_encoded, y_NNprob_alt)

print("\n" + "=" * 45)
print("  NEURAL NETWORK (Alternate Data) — EVALUATION RESULTS")
print("=" * 45)
print(f"  Accuracy  : {accuracy_nn_alt:.4f}  ← misleading for imbalanced!")
print(f"  Precision : {Precision_nn_alt:.4f}")
print(f"  Recall    : {Recall_nn_alt:.4f}   ← how many leavers caught?")
print(f"  F1-Score  : {F1_nn_alt:.4f}   ← PRIMARY metric")
print(f"  ROC-AUC   : {roc_auc_nn_alt:.4f}   ← discrimination ability")
print("=" * 45)

print("\nDetailed Classification Report:")
print(classification_report(y_alt_test_encoded, y_NNpred_alt, target_names=label_encoder.classes_))
```

```python
# Learning history plotting function
def plot_history(history):
  """Takes a keras model learning history and plots each metric"""

  metrics = history.history.keys()

  for metric in metrics:
      if not 'val' in metric:
        plt.plot(history.history[f'{metric}'], label=f'{metric}')
        if f'val_{metric}' in metrics:
          plt.plot(history.history[f'val_{metric}'], label=f'val_{metric}')
        plt.legend()
        plt.title(f'{metric}')
        plt.show()
```
